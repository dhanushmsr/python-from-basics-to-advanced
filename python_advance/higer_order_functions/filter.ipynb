```python
def even(l):
    if l%2==0:
        return True
print(list(filter(even,(9,13,5,6,8,14,68))))
```

```python
def odd(l):
    if l%2==0:
        return False
    else:
        return True
print(list(filter(odd,(9,13,5,6,8,14,68))))
```
