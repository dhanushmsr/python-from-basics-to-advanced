```python
import json
a={1:"one",2:'two',3:'three',4:'four'}
j1=json.dumps(a)
print(j1,type(j1))
```

```python
a={1:"one",2:'two',3:'three',4:'four'}
b=open("learn.json",'w')
json.dump(a,b)
b.close()
```

```python
a=[1,2,3,4,]
b=open("learn.json",'r+') #if i use w mode it will erase all data existing in the file and write the new so i use r+ mode
b.read()
json.dump(a,b) #but here since the python object is list previous data is dict so json format get error
b.close()
```

```python
a={1,2,3,4}
with open("learn.json", "r+") as b:
    json.dump(a,b) #Json doesn't support for set object
```

```python
a={'name':'dhanush','age':21,'education':'B.E'}
with open("learn.json","r+") as b:
    b.read()
    json.dump(a,b) #then when we used dump to write py object into json it must be one python object else causes error
```

```python
a=["dhanush","cse",21,"Coimbatore"]
b=open("learn.json","r+")
b.read()
json.dump(a,b) #similar error either it should have one list object with dictionary or one dictionary object with list  
```
