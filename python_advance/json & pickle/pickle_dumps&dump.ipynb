```python
import pickle
a={10,20,30}
b=open("learn.pkl","wb") #pickle.dump convert any python object into a binary
pickle.dump(a,b)
b.close()
```

```python
import pickle
a=10,20,30
print(pickle.dumps(a))
```

```python
a=b'\x80\x04\x95\t\x00\x00\x00\x00\x00\x00\x00K\nK\x14K\x1e\x87\x94.'
print(pickle.loads(a)) # convert any binary data into it's corresponding python object
```

```python
with open("learn.pkl","rb") as b:
    a=pickle.load(b) # reading from learn.pkl file
    print(a)
```
