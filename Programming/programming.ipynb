```python
#swapping numbers
a=int(input('a:'))
b=int(input('b:'))
c=int(input('c:'))
print(f"before swap a={a}\t b={b} \t c={c}")
a=a^b^c                         # a=a*b*c                                       # a=a+b+c                    # d=a
b=a^b^c                         # b=a//(b*c)                                    # b=a-b-c                    # a=b
c=a^b^c                         # c=a//(b*c)                                    # b=a-b-c                    # a=b
a=a^b^c                         # a=a//(b*c)                                    # c=a-b-c                    # b=c
                                #                                               # a=a-b-c                    # c=d
print(f"After swap a={a}\t b={b} \t c={c}")
```
