```python
#Number of prime numbers
def prime_number(n):
    count=0
    for i in range(1,n+1):
        if n%i==0:
            count+=1
    if count == 2:
        return True
    else:
        return False

def num_of_prime(num):
    l=[]
    val=2
    while len(l)<num:
        if prime_number(val):
            l.append(val)
        val+=1
    return l

num_of_prime(int(input("enter the number of prime number you want: ")))
        

```

```python
#Interger reversing
def reverse(n):
    sign=-1 if n<0 else 1
    n=abs(n) 
    res=0
    while n>0:
        rem=n%10
        res=res*10+rem
        n//=10
    return sign*res
print(reverse(-1240))
```

```python
#string reverser
def reverse(s):
    if s=="":
        return ""
    return reverse(s[1:])+s[0]
print(reverse("dhanush"))
```

```python
#Perfect number
n=int(input("n: "))
res=0
for i in range(1,n):
    if n%i==0:
        res=res+i
if res==n:
    print("perfect number") #6
else:
    print("not the perfect number")
```
