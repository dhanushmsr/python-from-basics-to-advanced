```python
#Number of prime numbers
def prime_number(n):
    count=0
    for i in range(1,n+1):
        if n%i==0:
            count+=1
    if count == 2:
        return True
    else:
        return False

def num_of_prime(num):
    l=[]
    val=2
    while len(l)<num:
        if prime_number(val):
            l.append(val)
        val+=1
    return l

num_of_prime(int(input("enter the number of prime number you want: ")))
        

```
