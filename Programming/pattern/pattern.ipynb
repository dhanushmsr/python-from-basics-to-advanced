```python
n=list(map(int,input('Enter the number of row and column with space').split()))
row=n[0]
col=n[1]
for i in range(row):
    for j in range(col):
        if i>j:
            print('*', end=' ')
        else:
            print(' ',end=' ')
    print()
```
