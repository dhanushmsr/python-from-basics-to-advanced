```python
t=(10,'py',True,[20,10],(4,6))
print(t)
(t[3][1])='sql' #we can modify the list inside the tuple but we cannot modify the tuple element
print(t)
```

```python
t=20,30,20,9.2,'py',34,20,'py'
print(t.index(9.2))
print(t.index('py'))
print(t.index("py",5))
print(t.count('py'))
```

```python
t=tuple(i for i in range(20) if i%2==0)
print(t)
```

```python
t=tuple(f"even({i})" if i%2==0  else f"odd({i})" for i in range(10))
print(t)
```
