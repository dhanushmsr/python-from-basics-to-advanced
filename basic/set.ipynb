```python
s={10,20,30,'py',40,10,(1,2)}
se={100,50,'py',30}
s.add(True)
print(s)
print(s.pop())
print(s.remove(10),s) #remove will rise error if element is not inside the set
s.discard(40) #discard will not rise error even element is not inside the set
print(s)
s1=s.copy()
print(s1)
print(s.union(se)) #union two set or set with other sequence and return new set
print(s.update(se),s) #update two set or set with other sequence and update in primary set ex., here s is primary set
```

```python
s={10,20,30,'py',40,10,(1,2)}
se={100,50,'py',30}
print(s.intersection(se)) #return only the common element in both set or set with other sequence
s.intersection_update(se)
print(s)
```

```python
s={10,20,30,'py',40,10,(1,2)}
se={100,50,'py',30}
print(s.difference(se)) #difference method check the common element in both set and return only the different primary set element
```

```python
s={10,20,30,'py',40,10,(1,2)}
se={100,50,'py',30}
s.difference_update(se) #in this method it find the difference and remove that and only the primary set element get updated on the primary set itself
print(s)
```

```python
s={10,20,30,'py',40,10,(1,2)}
print(s) #set are unordered data and remove the duplicates
```

```python
s={10,20,30,'py',40,10,(1,2)}
se={100,50,'py',30}
print(s.symmetric_difference(se)) #return all different element on both set or set with other ommit only the common part
```

```python
s={10,20,30,'py',40,10,(1,2)}
se={10,20,30,'py',40,10,(1,2)}
print(s.symmetric_difference(se))
```

```python
s={10,20,30,'py',40,10,(1,2)}
se={100,'py',50,30}
s.symmetric_difference_update(se) # it update the common differenc element on both set on the primary set
print(s)
```

```python
s={10,20,30,'py',40,10,(1,2)}
se={100,'py',50,30}
print(s.isdisjoint(se)) #return bool value if it have same element False if no same element then True
```

```python
s={10,20,30,40,60}
print(s.issubset([10,20]))
print(s.issubset((10,20)))
print(s.issubset('py'))
print({10,20}.issubset(s))
```
